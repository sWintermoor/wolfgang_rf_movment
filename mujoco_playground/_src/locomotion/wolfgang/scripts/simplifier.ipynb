```python
import xml.etree.ElementTree as ET
```

```python
tree = ET.parse('country_data.xml')
root = tree.getroot()
print(root.tag, root.attrib)
for child in root:
    print(child.tag, child.attrib)
    for subchild in child:
        print(subchild.tag, subchild.attrib, subchild.text)
```
