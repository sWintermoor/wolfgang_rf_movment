```python
import onnx
import numpy as np
import onnxruntime as ort
```

```python
# Loading the ONNX model
model = onnx.load("wolfgang_walk_ppo.onnx")
```

```python
onnx.checker.check_model(model)
print("The model is valid!")
```

```python
OBSERVATION_SIZE = 78

observation = np.zeros((1, OBSERVATION_SIZE), dtype=np.float32)
session = ort.InferenceSession("wolfgang_walk_ppo.onnx")

input_names = [input.name for input in session.get_inputs()]
print("Input name:", input_names)
```

```python
output = session.run(None, {"in_0": observation})
print("Model output:", output)
print("Output shape:", output[0].shape)
```
